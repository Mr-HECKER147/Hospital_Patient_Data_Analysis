# Hospital Data Overview
Load the hospital dataset and show key data summaries without plotting duplicate visualizations.


```python
import numpy as np
import pandas as pd

df = pd.read_csv("hospital.csv")
print(df.head())
```

```python
# Highlight the highest-cost patients
print(df[df['Total_Cost'] > 6000][['Patient_ID', 'Age', 'Condition', 'Total_Cost']].head(30))
```

```python
# Basic statistics
print("=== Basic Statistics ===")
print(df.describe())
print("=== Data Info ===")
print(df.info())

# Patient counts by condition and outcome
print("=== Patients by Condition ===")
print(df["Condition"].value_counts().head(10))
print("=== Outcome Distribution ===")
print(df["Outcome"].value_counts().head(10))

# Age distribution summary
print("=== Age Distribution ===")
print(f"Mean Age: {df['Age'].mean():.2f}")
print(f"Median Age: {df['Age'].median():.2f}")
print(f"Age Range: {df['Age'].min()} - {df['Age'].max()}")

# Top states by patient count
print("=== Top 10 States by Patient Count ===")
print(df["Patient_State"].value_counts().head(10))
```
