# Hospital Financial and Readmission Analysis
Analyze treatment cost, length of stay, insurance usage, and readmission behavior.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("hospital.csv")
```

```python
plt.figure(figsize=(15, 8))
sns.scatterplot(data=df, x="Length_of_Stay", y="Total_Cost", hue="Condition", alpha=0.7)
plt.title("Relationship Between Stay Duration and Total Cost")
plt.show()
```

```python
avg_cost = df.groupby("Condition")["Total_Cost"].mean().sort_values()
plt.figure(figsize=(12, 6))
avg_cost.plot(kind="barh", color="teal")
plt.title("Average Treatment Cost per Condition")
plt.xlabel("Average Cost")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(
    data=df,
    x="Satisfaction",
    hue="Satisfaction",
    legend=False,
    palette="RdYlGn"
)
plt.title("Patient Satisfaction Ratings (1-5)")
plt.show()
```

```python
readmission_rate = df["Readmission"].value_counts(normalize=True) * 100
print("Readmission Rate (%) by Overall Status:", readmission_rate)

insurance_stats = df.groupby("Condition")["Insurance_Claimed"].value_counts(normalize=True).unstack() * 100
plt.figure(figsize=(12, 6))
insurance_stats["Yes"].sort_values().plot(kind="barh", color="orchid")
plt.title("Percentage of Insurance Claims per Condition")
plt.xlabel("Claims (%)")
plt.show()
```
